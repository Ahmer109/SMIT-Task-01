```python
a = input("Enter your name: ")
b = input("Enter a number: ")
c = input("Enter a sentence: ")

print("\n--- String Methods ---")

print("Upper:", a.upper())
print("Lower:", a.lower())
print("Title:", c.title())
print("Capitalize:", c.capitalize())
print("Swapcase:", a.swapcase())

print("Strip:", c.strip())
print("Lstrip:", c.lstrip())
print("Rstrip:", c.rstrip())

print("Replace:", c.replace("Python", "Java"))
print("Split:", c.split())
print("Join:", "-".join(c.split()))

print("Find:", a.find(a[0]))
print("Index:", a.index(a[0]))
print("Count:", a.count(a[0]))

print("Startswith:", a.startswith(a[0]))
print("Endswith:", a.endswith(a[-1]))

print("Isalpha:", a.isalpha())
print("Isdigit:", b.isdigit())
print("Isalnum:", a.isalnum())
print("Isspace:", c.isspace())

print("Islower:", a.islower())
print("Isupper:", a.isupper())
print("Istitle:", c.istitle())

print("Center:", a.center(20, "*"))
print("Ljust:", a.ljust(20, "-"))
print("Rjust:", a.rjust(20, "-"))

print("Zfill:", b.zfill(10))

print("Partition:", c.partition(" "))
print("Rpartition:", c.rpartition(" "))

print("Casefold:", a.casefold())
```
